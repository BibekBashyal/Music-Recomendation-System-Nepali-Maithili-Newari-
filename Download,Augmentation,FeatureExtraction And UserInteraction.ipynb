```python
import os
import random
import librosa
import soundfile as sf
from audiomentations import Compose, AddGaussianNoise, TimeStretch, PitchShift, Shift

# Paths
input_folder = "AI/songs"
output_folder = "AI/augmented_songs"
os.makedirs(output_folder, exist_ok=True)

# Define multiple augmentation strategies
def get_random_augment():
    pipelines = [
        Compose([AddGaussianNoise(min_amplitude=0.001, max_amplitude=0.01, p=1.0)]),
        Compose([TimeStretch(min_rate=0.9, max_rate=1.1, p=1.0)]),
        Compose([PitchShift(min_semitones=-2, max_semitones=2, p=1.0)]),
        Compose([Shift(min_shift=-0.2, max_shift=0.2, p=1.0)]),
        Compose([
            AddGaussianNoise(p=0.5),
            PitchShift(p=0.5),
            TimeStretch(p=0.5),
            Shift(p=0.5)
        ])
    ]
    return random.choice(pipelines)

# Process each song
for file in os.listdir(input_folder):
    if file.endswith(".wav") or file.endswith(".mp3"):
        path = os.path.join(input_folder, file)
        y, sr = librosa.load(path, sr=None)

        for i in range(20):  # 20 augmentations per song
            augment = get_random_augment()
            y_aug = augment(samples=y, sample_rate=sr)
            new_filename = f"{os.path.splitext(file)[0]}_aug{i}.wav"
            sf.write(os.path.join(output_folder, new_filename), y_aug, sr)

print("✅ Successfully generated 400 augmented audio files.")
```

```python
import pandas as pd
import numpy as np

feature_data = []

for file in os.listdir(output_folder):
    if file.endswith(".wav"):
        path = os.path.join(output_folder, file)
        y, sr = librosa.load(path, sr=None)

        # Extract features
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13).mean(axis=1)
        chroma = librosa.feature.chroma_stft(y=y, sr=sr).mean(axis=1)
        tempo, _ = librosa.beat.beat_track(y=y, sr=sr)

        feature_data.append({
            "filename": file,
            "tempo": tempo,
            **{f"mfcc{i+1}": mfcc for i, mfcc in enumerate(mfccs)},
            **{f"chroma{i+1}": c for i, c in enumerate(chroma)}
        })

# Save to CSV
df = pd.DataFrame(feature_data)
df.to_csv("AI/song_features.csv", index=False)
print("✅ Feature extraction completed and saved to 'song_features.csv'")
```

```python
import os
import pandas as pd
import random

# Load song filenames from folder
folder_path = "AI/augmented_songs"
song_files = [f for f in os.listdir(folder_path) if f.endswith(".wav") or f.endswith(".mp3")]

# Create DataFrame manually
df = pd.DataFrame(song_files, columns=["filename"])

# Create user interaction data
users = [f"user{i+1}" for i in range(50)]
songs = df["filename"].tolist()
interaction_data = []

for user in users:
    liked = random.sample(songs, k=random.randint(5, 15))
    for song in liked:
        interaction_data.append({
            "user": user,
            "song": song,
            "rating": random.randint(3, 5)
        })

# Save to CSV
df_interactions = pd.DataFrame(interaction_data)
df_interactions.to_csv("AI/user_song_interactions.csv", index=False)
print("✅ Simulated user interaction data saved to 'AI/user_song_interactions.csv'")
```
